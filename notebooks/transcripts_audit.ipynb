# Reference vs Whisper Generated Transcripts Audit

```python
import pandas as pd
import torch
from collections import Counter
from itertools import chain
import jiwer
from jiwer import wer
import matplotlib.pyplot as plt
import torch
```

```python
reference_inputs = torch.load('embeddings/linguistic/distilbert_reference/distilbert_reference_train.pt')
reference_uids = pd.Series([i[0] for i in reference_inputs], name = 'SNo')
train_manifest = pd.read_csv('results/artifacts/train_manifest.csv')
train_manifest = train_manifest[['SNo', 'Reference_Transcript', 'Label']]
merged = pd.merge(reference_uids, train_manifest, on='SNo', how='inner')
merged = merged.drop_duplicates()
reference = merged[['SNo', 'Reference_Transcript', 'Label']].reset_index(drop=True)

whisper_raw = pd.read_csv('results/artifacts/whisper_transcripts_train.csv')[['SNo', 'Whisper_Transcript', 'Label']].reset_index(drop = True)
whisper = whisper_raw.copy()
whisper['Whisper_Transcript'] = (whisper['Whisper_Transcript'].fillna('').astype(str).str.strip())
```

```python
total_len = len(reference)
print("total samples: ", total_len)
```

### Number of Usable Transcripts

```python
reference_usable = reference['Reference_Transcript'].str.strip().ne("").sum()
whisper_usable = whisper['Whisper_Transcript'].str.strip().ne("").sum()

print(f"number of usable samples in reference: {reference_usable}")
print(f"number of usable samples in whisper: {whisper_usable}")

if whisper_usable < reference_usable:
    print(f"\nnumber of missing transcripts in whisper: {reference_usable - whisper_usable}")
```

### Coverage & Missing Rates

```python
reference_cr = (reference_usable / total_len) * 100
whisper_cr = (whisper_usable / total_len) * 100

print(f"coverage rate of reference: {reference_cr:.2f}")
print(f"coverage rate of whisper: {whisper_cr:.2f}")
```

```python
reference_mr = 100 - reference_cr
whisper_mr = 100 - whisper_cr

print(f"missing rate of reference: {reference_mr:.2f}")
print(f"missing rate of whisper: {whisper_mr:.2f}")
```

### Emotion Coverage by Whisper

```python
whisper_emotions = whisper['Label']
whisper_emo_cov = dict(Counter(whisper_emotions))
print("emotion coverage by whisper:\n", whisper_emo_cov)

reference_emotions = merged['Label'].reset_index(drop=True)
reference_emo_cov = dict(Counter(reference_emotions))
print("actual emotion distribution:\n", reference_emo_cov)

print()
for emotion in whisper_emo_cov.keys():
    miss_rate = (abs((whisper_emo_cov[emotion] - reference_emo_cov[emotion])) / reference_emo_cov[emotion]) * 100
    print(f"miss rate for {emotion}: {miss_rate:.2f}% ({whisper_emo_cov[emotion]}/{reference_emo_cov[emotion]})")
```

### Average Transcript Length

```python
ref_total_len = sum([len(i) for i in reference['Reference_Transcript']])
ref_avg_len = ref_total_len / len(reference)
whis_total_len = sum([len(i) for i in whisper['Whisper_Transcript']])
whis_avg_len = whis_total_len / len(whisper)

print(f"average length of reference transcripts: {ref_avg_len:.2f}")
print(f"average length of whisper transcripts: {whis_avg_len:.2f}")
print(f"difference in raw characters: {ref_total_len - whis_total_len}")
print(f"difference in averages: {ref_avg_len - whis_avg_len:.2f}")
```

### Word Counts

```python
ref_words_raw = [i.split(" ") for i in reference['Reference_Transcript']]
ref_words = len(list(chain.from_iterable(ref_words_raw)))

whis_words_raw = [i.split(" ") for i in whisper['Whisper_Transcript']]
whis_words = len(list(chain.from_iterable(whis_words_raw)))

print(f"total words in reference transcripts: {ref_words}")
print(f"total words in whisper transcripts: {whis_words}")
print(f"difference: {ref_words - whis_words}")
```

## Word Error Rate (WER)

```python
wers = []
ref_whis = pd.merge(reference, whisper, how = 'inner', on = 'SNo').drop_duplicates()

for row in ref_whis.itertuples():
    wers.append(wer(row.Reference_Transcript, row.Whisper_Transcript))

avg_wer = (sum(wers) / len(wers)) * 100
print(f"average word error rate: {avg_wer:.2f}%")
```

### Word Error Rate by Emotion

```python
emotion_wers = {}
for emotion in whisper_emo_cov.keys():
    temp_df = ref_whis[ref_whis['Label_x'] == emotion]
    wers = []
    for row in temp_df.itertuples():
        wers.append(wer(row.Reference_Transcript, row.Whisper_Transcript))

    avg_wer = (sum(wers) / len(wers)) * 100
    emotion_wers[emotion] = avg_wer

    print(f"word error rate for {emotion}: {avg_wer:.2f}%")

plt.pie(emotion_wers.values(), labels = emotion_wers.keys())
plt.title("WER by Emotion Distribution")
plt.show()
```

### Detailed Analysis of WER

```python
sample_df = ref_whis.groupby('Label_x', group_keys = False).apply(lambda x: x.sample(n = 3, random_state = 42))
print("### sanity check (len of sample_df must be 3 * 7 = 21): ", len(sample_df))
print("analysis of 21 random samples: ")
for row in sample_df.itertuples():
    output = jiwer.process_words(row.Reference_Transcript, row.Whisper_Transcript)
    print(f"\n--- analysis for SNo {row.SNo} --- ")
    print(f"reference: {row.Reference_Transcript} / whisper: {row.Whisper_Transcript}")
    print(f"emotion: {row.Label_y}")
    print(f"substitutions: {output.substitutions}, insertions: {output.insertions}, deletions: {output.deletions}, correct words: {output.hits}")
```

### Classification Errors vs WER

```python
from src.models.mlp import MultiLayerPerceptron
from src.models.dataloader import SpeechEmbeddingsDataset
from torch.utils.data import DataLoader
```

```python
# 1. Find the classification results of Whisper-only model

MODEL_PATH =  'results/models/baselines/base_transcripts.pt'
EMBEDDINGS_PATH = 'embeddings/linguistic/distilbert_transcript/distilbert_transcripts_test.pt'

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = MultiLayerPerceptron()
state_dict = torch.load(MODEL_PATH, map_location=device)
model.load_state_dict(state_dict)
model = model.to(device)

test_inputs = torch.load(EMBEDDINGS_PATH)
uids = [i[0] for i in test_inputs]
embeddings = torch.stack([i[1] for i in test_inputs])
labels = [i[2] for i in test_inputs]

test_dataset = SpeechEmbeddingsDataset(embeddings, labels)
test_loader = DataLoader(test_dataset, batch_size=100, shuffle=False)

model.eval()
all_preds, all_trues = [], []
with torch.no_grad():
    for batch_embeddings, batch_labels in test_loader:
        batch_embeddings = batch_embeddings.to(device)
        batch_labels = batch_labels.to(device)

        outputs = model(batch_embeddings)
        preds = torch.argmax(outputs, dim = 1).detach().cpu().numpy()
        
        all_preds.extend(preds)
        all_trues.extend(batch_labels.cpu().numpy())
```

```python
# 2. Determine Correct/Incorrect Classification

predictions = [i.item() for i in all_preds]
labels = [i.item() for i in all_trues]
classified = [1 if predictions[i] == labels[i] else 0 for i in range(len(predictions))]
```

```python
# 3. Initializing Test DataFrames
reference_test_inputs = torch.load('embeddings/linguistic/distilbert_reference/distilbert_reference_test.pt')
reference_test_uids = pd.Series([i[0] for i in reference_test_inputs], name = 'SNo')
test_manifest = pd.read_csv('results/artifacts/test_manifest.csv')
test_manifest = test_manifest[['SNo', 'Reference_Transcript', 'Label']]
merged_test = pd.merge(reference_test_uids, test_manifest, on='SNo', how='inner')
merged_test = merged_test.drop_duplicates()
reference_test = merged_test[['SNo', 'Reference_Transcript', 'Label']].reset_index(drop=True)

whisper_test_raw = pd.read_csv('results/artifacts/whisper_transcripts_test.csv')[['SNo', 'Whisper_Transcript', 'Label']].reset_index(drop = True)
whisper_test = whisper_test_raw['Whisper_Transcript'].astype(str).str.strip().replace(['', 'nan', 'NaN', 'None'], None)
whisper_test = whisper_test_raw.dropna()

ref_whis_test = pd.merge(reference_test, whisper_test, how = 'inner', on = 'SNo').drop_duplicates()

print("### sanity check")
print("number of samples classified by model: ", len(classified))
print("number of samples in the reference-whisper merged df: ", len(ref_whis_test))
```

```python
ref_whis_test['Classified'] = classified
compare = ref_whis_test[['Reference_Transcript', 'Whisper_Transcript', 'Classified']]
classified_wers = {}
for i in range(2):
    temp = compare[compare['Classified'] == i]
    wers = []
    for row in temp.itertuples():
        wers.append(wer(row.Reference_Transcript, row.Whisper_Transcript))

    avg_wer = sum(wers) / len(wers)
    classified_wers[i] = avg_wer

    print(f"word error rate for {i}: {avg_wer:.4f}")

plt.pie(classified_wers.values(), labels = classified_wers.keys())
plt.title("WER by Classification")
plt.show()
```

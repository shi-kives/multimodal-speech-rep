```python
import pandas as pd
import torch
import math
import numpy as np
```

```python
def run_inspection(embeddings: list, modality: str):
    print(f"\n---- DATA INSPECTION FOR {modality} -----\n")

    print(f"number of samples: {len(embeddings)}")
    print(f"shape of one embedding: {embeddings[0][1].shape}\n")

    df = pd.DataFrame(embeddings, columns = ['SNo', 'Embeddings', 'Label'])
    print(f"null values along any axis?:\n{df.isna().sum()}\n")

    print(f"infinite values along col sample ids: {np.isinf(df['SNo']).sum()}")
    tensor_col = [i[1] for i in embeddings]
    stacked_emb = torch.stack(tensor_col)
    print(f"infinite values among embeddings: {torch.isinf(stacked_emb).sum()}\n")   

    print(f"number of unique emotion classes: {df['Label'].nunique()}\n")
    print(f"number of unique sample ids: {df['SNo'].nunique()}")
    print(f"duplicate sample ids: {df['SNo'].duplicated().sum()}\n")
    print(f"class distribution for emotions: {df.groupby('Label')['SNo'].count()}\n")

    manifest = pd.read_csv('results/artifacts/train_manifest.csv')
    result = pd.merge(df, manifest, on='SNo', how='inner')
    print(f"matching samples: {len(result)}")
```

```python
acoustic = torch.load('embeddings/acoustic/wavlm_features.pt')
distilbert_transcripts = torch.load('embeddings/linguistic/distilbert_transcripts.pt')
distilbert_reference = torch.load('embeddings/linguistic/distilbert_reference.pt')
```

```python
run_inspection(acoustic, 'ACOUSTIC')
```

```python
run_inspection(distilbert_reference, 'DISTILBERT-REFERENCE')
```

```python
run_inspection(distilbert_transcripts, 'DISTILBERT-TRANSCRIPTS')
```

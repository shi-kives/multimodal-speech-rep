## Conclusion

**Strongest Observations:**
1. Most improvement was found post class weighting, where balanced accuracy and macro f1 scores shot up by 5% and 1% respectively, in direct fusion.
2. Worst performance was found in projected fusion with projection dimensions 256 over the baseline 128, balanced accuracy and macro f1 score dropping by 5% each approximately.
3. Learning Rate = 0.0003 showed worse performance than Learning Rate = 0.0010, metrics dipping by ~3%, and ~2.5%.
4. Dropout p = 0.1 also showed poor performance, with balanced accuracy and macro-f1 score dropping by ~2.5% and ~2% each.

**Common Observations:**
1. Early stopping showed no improvement in the performance of the model with both the min delta values, the exact same scenario took place during validation.
2. Hidden dimensions = 64 dropped the performance in testing by ~2% each for balanced accuracy and hidden dimensions, which occured during validation also.

**Other Observations:**
1. Hyperparameter values Weight Decay = 0.01 and 0.001 showed moderate but useful improvements
2. Whereas, Dropout = 0.3 showed a moderate performance drop.
